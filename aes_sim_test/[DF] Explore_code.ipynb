```python
def add_comment(pattern, position_file,instructions_list, comment):
    positions_list= load_list(position_file)
    ranges = []
    for pos in positions_list:
        ranges.extend(range(pos, pos + len(pattern)))
    for j in range(len(positions_list)):
        start_pos=positions_list[j]
        for j in range(len(pattern)):
            instructions_list[start_pos + j]= comment
    return instructions_list
```
